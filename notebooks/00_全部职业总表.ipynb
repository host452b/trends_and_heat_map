# 全球职业发展指数 — 职业总表

**Global Career Development Index — Occupation Master Table**

1,200 个职业 · 每行综合45国数据 · 34维全球均分 · 12区域综合指数对比

- **34维评分列**: 各国均值，红→黄→绿色阶
- **区域综合指数**: 按12个区域分组的composite_index均值

评分色阶：红色(低分0) → 黄色(中等5) → 绿色(高分10)  
口碑方差：绿色(稳定0) → 红色(分化5)  
趋势：红色(暴跌-5) → 白色(持平0) → 绿色(暴涨+5)
